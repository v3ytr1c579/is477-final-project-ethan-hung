## RESEARCH QUESTION: How does a person's role in a crash (driver, passenger, pedestrian, bicyclist) affect injury severity in Chicago traffic accidents?

```python
filtered = merged[merged["PERSON_TYPE"].isin(["DRIVER", "PASSENGER", "PEDESTRIAN", "BICYCLE"])]

injury_by_person = (
    filtered.groupby("PERSON_TYPE")["INJURY_CLASSIFICATION"]
    .value_counts(normalize=True)
    .unstack()
)

injury_by_person.plot(kind="bar", stacked=True)

plt.title("Injury Severity Distribution by Person Type")
plt.ylabel("Proportion")
plt.xlabel("Person Type")

plt.legend(title="Injury Classification", bbox_to_anchor=(1.05, 1))
plt.xticks(rotation=45)

plt.savefig("../outputs/injury_by_person_type.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

## FINDINGS

The visualization above shows clear differences in the severity of an injury depending on a person's role in a crash. For instance, drivers and passengers report virtually no injury; this indicates that individuals inside vehicles experience relatively low injury severity in most crashes. However, bicyclists and pedestrians on the other hand show much higher proportions of injuries; pedestrians specifically have the largest share of non-incapacitating and incapacitating injuries and are the only group with a noticeable proportion of fatal injury cases. Drivers and passengers show near-zero fatal injury proportions by comparison. These results overall suggest that individuals outside vehicles are considerably more vulnerable to injury during traffic accidents.
